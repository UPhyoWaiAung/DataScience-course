```python
import pandas as pd

# Define path and column headers
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/adult/adult.data"
columns = [
    'age', 'workclass', 'fnlwgt', 'education', 'education-num',
    'marital-status', 'occupation', 'relationship', 'race', 'gender',
    'capital-gain', 'capital-loss', 'hours-per-week', 'native-country', 'income'
]

# Load and subset data
data = pd.read_csv(url, header=None, names=columns, skipinitialspace=True)
selected_cols = ['age', 'workclass', 'education', 'gender', 'hours-per-week', 'occupation', 'income']
data = data[selected_cols]

# Convert categorical columns to dummy numeric variables for machine learning
data_dummies = pd.get_dummies(data, drop_first=True)
display(data_dummies.head())
```
